```python
import cv2
import matplotlib.pyplot as plt
img_bgr = cv2.imread("image.jpeg")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
img_bgr = cv2.imread("image.jpeg")
plt.imshow(img_bgr)
plt.title("Image using Matplotlib")
plt.show()
```

```python
from skimage import io
import matplotlib.pyplot as plt
img_ski = io.imread("image.jpeg")
plt.imshow(img_ski)
plt.title("Image using Scikit-Image")
plt.show()
```
