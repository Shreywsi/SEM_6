```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)

# function
def digitize(image, sampling_interval, L):
    sampled = image[::sampling_interval, ::sampling_interval]
    step = 256 // L
    quantized = (sampled // step) * step
    return quantized

sampling_interval = 2
L = 4   
quantized_img = digitize(img, sampling_interval, L)

print("Original Mean:", np.mean(img))
print("Quantized Mean:", np.mean(quantized_img))

plt.hist(img.ravel(), bins=256)
plt.title("Original Histogram")
plt.show()

plt.hist(quantized_img.ravel(), bins=L)
plt.title("Quantized Histogram")
plt.show()
```
