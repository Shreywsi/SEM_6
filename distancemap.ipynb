```python
import numpy as np
import matplotlib.pyplot as plt

binary_image = np.zeros((20, 20))
binary_image[5:15, 5:15] = 1

source = (10, 10)

def d4_distance_map(shape, source):
    s, t = source
    dist_map = np.zeros(shape)
    for x in range(shape[0]):
        for y in range(shape[1]):
            dist_map[x, y] = abs(x - s) + abs(y - t)
    return dist_map

def d8_distance_map(shape, source):
    s, t = source
    dist_map = np.zeros(shape)
    for x in range(shape[0]):
        for y in range(shape[1]):
            dist_map[x, y] = max(abs(x - s), abs(y - t))
    return dist_map

d4_map = d4_distance_map(binary_image.shape, source)
d8_map = d8_distance_map(binary_image.shape, source)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(binary_image, cmap='gray')
plt.scatter(source[1], source[0])

plt.subplot(1, 3, 2)
plt.imshow(d4_map)
plt.scatter(source[1], source[0])

plt.subplot(1, 3, 3)
plt.imshow(d8_map)
plt.scatter(source[1], source[0])

plt.show()
```
