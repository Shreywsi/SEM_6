```python
import random
import numpy as np

def mp_neuron(inputs, threshold):
    return 1 if sum(inputs) > threshold else 0
```

```python
#Generate Synthetic Data
def generate_data(n=500):
    X = []
    y = []

    for i in range(n):
        age = random.randint(20, 60)              # Age
        salary = random.randint(20000, 100000)    # Salary
        family = random.randint(1, 6)             # Family members

        inputs = [age, salary, family]
        X.append(inputs)

        # True label (synthetic rule)
        label = 1 if (age + salary + family) > 80000 else 0
        y.append(label)

    return np.array(X), np.array(y)

X, y = generate_data(500)
```

```python
#Train-Test Split
split_index = int(0.7 * len(X))

X_train, X_test = X[:split_index], X[split_index:]
y_train, y_test = y[:split_index], y[split_index:]
```

```python
# Training = Compute threshold
threshold = np.mean([sum(x) for x in X_train])

# Testing
correct = 0
for inputs, label in zip(X_test, y_test):
    prediction = mp_neuron(inputs, threshold)
    if prediction == label:
        correct += 1

accuracy = correct / len(y_test)

print("Threshold:", threshold)
print("Test Accuracy:", accuracy)

```
