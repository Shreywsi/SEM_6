```python
def policy_iteration(states, actions, transition, reward, gamma=0.9):
    
    # Initialise policy (skip terminal state)
    policy = {s: actions[s][0] for s in states if actions[s]}
    values = {s: 0 for s in states}

    while True:
        #  Policy Evaluation
        for s in states:
            if not actions[s]:   # skip terminal (G)
                continue
            a = policy[s]
            values[s] = reward[s][a] + gamma * values[transition[s][a]]

        #  Policy Improvement
        new_policy = {}
        for s in states:
            if not actions[s]:   # skip terminal
                continue
            new_policy[s] = max(actions[s],key=lambda a: reward[s][a] + gamma * values[transition[s][a]])

        # Stop if no change
        if new_policy == policy:
            break

        policy = new_policy

    return values, policy
```

```python
states = ['A', 'B', 'C', 'D', 'E', 'G']

actions = {
    'A': ['right', 'down'],
    'B': ['left', 'right', 'down'],
    'C': ['left', 'down'],
    'D': ['up', 'right'],
    'E': ['up', 'left', 'right'],
    'G': []   
}

transition = {
    'A': {'right': 'B', 'down': 'D'},
    'B': {'left': 'A', 'right': 'C', 'down': 'E'},
    'C': {'left': 'B', 'down': 'G'},
    'D': {'up': 'A', 'right': 'E'},
    'E': {'up': 'B', 'left': 'D', 'right': 'G'},
    'G': {}   
}

reward = {
    'A': {'right': 0, 'down': 0},
    'B': {'left': 0, 'right': 1, 'down': 0},
    'C': {'left': 0, 'down': 2},
    'D': {'up': 0, 'right': 0},
    'E': {'up': 0, 'left': 0, 'right': 3},  # best path to G
    'G': {}   
}

V, policy = policy_iteration(states, actions, transition, reward)

print("Values:", V)
print("Policy:", policy)
```
