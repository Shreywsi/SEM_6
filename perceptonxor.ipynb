```python
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,1,1,0])
w = np.zeros(2)
b = 0
lr = 0.1
epochs = 10
for _ in range(epochs):
    for i in range(len(X)):
        linear_output = np.dot(w, X[i]) + b
        y_pred = 1 if linear_output >= 0 else 0
        error = y[i] - y_pred
        w += lr * error * X[i]
        b += lr * error
for i in X:
    y_pred = 1 if np.dot(w, i) + b >= 0 else 0
    print(f"Input: {i}, Predicted: {y_pred}")
```
